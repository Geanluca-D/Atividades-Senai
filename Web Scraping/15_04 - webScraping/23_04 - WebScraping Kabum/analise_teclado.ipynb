```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn

df = pd.read_excel('Top5_Marcas.xlsx')
```

```python
#Média dos valores das marcas que mais tem opções

top5 = ['Akko', 'Kuromori', 'Motospeed', 'Redragon', 'Risemode']

tabela = df

mediaTop5 = tabela.groupby('Marca')['Preco_ajustado'].mean()


print(mediaTop5)
# print(tabela[['Preco_ajustado', 'Marca']])
```

```python
#Gerando o gráfico para a análise

plt.figure(figsize=(10,5))
mediaTop5.plot(kind='bar', color='purple')
plt.title('Média dos preços de teclados mecânicos (c/ switch azul)')
plt.xlabel('Marcas')
plt.ylabel('Valor médio dos valores')
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.show()
```
