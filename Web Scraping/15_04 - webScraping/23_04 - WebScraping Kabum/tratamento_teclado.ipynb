```python
import pandas as pd

df = pd.read_excel('TecladosKabum.xlsx')
print(df)
```

```python
df['Preco_ajustado'] = (
    df['Preço']
    .str.replace('R\$', '', regex=True)
    .str.replace('.', '', regex=False)
    .str.replace(',', '.', regex=False)
    .astype(float)
)

print(df)
```

```python
marcas = ['redragon', 'kbm', 'rise mode', 'dazz', 'warrior', 'fallen', 'akko', 'draxen', 'hyperx',
          'kirin', 'logitech', 'kuromori', 'marvo', 'msi', 'rapoo', 'razer', 'hyrax', 'fantech', 'motospeed',
          'orion', 'c3tech', 'kross elegance', 't-dagger', 'aplus tech', 'gamdias', 'sate', 'hoopson',
          'fortrek', 'xzone', 'bright', 'ditix', 'evolut', 'randgrid', 'hp', 'oex', 'steelseries', 'pcyes',
          ]

def extrair_marca(texto):
    texto = texto.lower()
    for marca in marcas:
        if marca in texto:
            return marca.capitalize()
    return 'Desconhecida'

df['Marca'] = df['Teclados mecânicos'].apply(extrair_marca)
```

```python
df['Custo'] = pd.cut(df['Preco_ajustado'], bins = [0, 299, 499, 9999], labels = ['Baixo custo', 'Médio custo', 'Alto custo'])

df.to_excel('Teclados_Tratados.xlsx', index=False)
```
