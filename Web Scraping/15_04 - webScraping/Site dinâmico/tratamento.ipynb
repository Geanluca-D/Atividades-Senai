```python
import pandas as pd

df = pd.read_excel('CadeirasKabum.xlsx')
print(df)
```

```python
cores = ['preto', 'preta', 'black', 'branco', 'branca', 'off white', 'vermelho', 'vermelha',
         'laranja', 'amarelo', 'amarela', 'verde', 'roxo', 'roxa', 'azul', 'rosa', 'grafite',
         'marrom', 'cinza', 'red', 'white', 'uva', 'navy blue', 'geada', 'ciano', 'yellow', 'ouro',
         'camuflado', 'camuflada', 'silver', 'blue', 'vinho', 'orange', 'camo']

def extrair_cor(texto):
    texto = texto.lower()
    for cor in cores:
        if cor in texto:
            return cor
    return "Desconhecida"

df['Cor Principal'] = df['Título'].apply(extrair_cor)

df[['Título', 'Cor Principal']].head(10)
```

```python
print(df[df['Cor Principal']=='Desconhecida'])

df[df['Cor Principal']=='Desconhecida'].to_excel('Cores.xlsx', index=False)
```

```python
contagem = df['Cor Principal'].value_counts()[2]

print(f'Quantidade de cadeiras sem cor expecificada: {contagem}')
```
