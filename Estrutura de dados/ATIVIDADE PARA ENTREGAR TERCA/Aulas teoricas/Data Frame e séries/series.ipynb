```python
import numpy as np
import pandas as pd

#Criar uma serie a partir de uma lista
s = pd.Series([1, 2, 3, 4, 5])

print(s)
```

```python
s = pd.Series([1,2,3,4,5], index=['a', 'b','c', 'd', 'e'])

print(s)
```

```python
s= pd.Series(np.random.uniform(0, 10), index=[f'{i}' for i in range(1,10)])

print(s)
```
