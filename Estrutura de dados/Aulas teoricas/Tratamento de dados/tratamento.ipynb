```python
import pandas as pd

df = pd.read_excel('dados_ausentes.xlsx')
print(df)
```

```python
#Corrigir erros de formatação (maiúsculo, minúsculo)

df['Cidade'] = df['Cidade'].str.title()
#Pode usar 'captalize' no lugar de 'title', 'title' deixa todas as primeiras letras das palavras escritas em maiúscula
#'capitalize' deixa apenas a primeira letra da primeira palavra

df['Nome'] = df['Nome'].str.capitalize()

print(df)
```

```python
#Substituição de valores categoricos

df['Cidade'] = df['Cidade'].str.lower().replace({'são paulo': 'SP', 'rio de janeiro': 'RJ', 'Rio de Janeiro' : 'RJ', 'sp' : 'SP'}) #Substitui esses valores
print(df)
```

```python
#Identificar os tipos de dados das colunas

coluna_string = df.select_dtypes(include=['string', 'object']).columns.to_list()
print(coluna_string)
```

```python
#Identificar os tipos de dados das colunas

coluna_float = df.select_dtypes(include=['float']).columns.to_list()
print(coluna_float)
```

```python
#Substituir idades ausentes pela média
df['Idade'].fillna(df['Idade'].mean(), inplace = True)
print(df)
```

```python
#Converter os dados de idade para int
df['Idade'] = df['Idade'].astype(int)
print(df)
```

```python
#Substitui os valores vazios pelo valor a cima
df['Idade'] = df['Idade'].ffill()

#Substitui os valores vazios pelo valor a baixo
df['Idade'] = df['Idade'].bfill()
```

```python
#Criar uma nova coluna: faixa etária

df['Faixa Etária'] = pd.cut(df['Idade'], bins = [0, 18, 30, 50, 100], labels=['Jovem', 'Adulto', 'Meia-Idade', 'Idoso'])
#O cut separa alguma lista na faixa definida no bins e em labels classifica o nome das faixas

print(df)
```

```python
#Remover duplicatas

df = df.drop_duplicates(subset=['Nome'])
print(df)
```
