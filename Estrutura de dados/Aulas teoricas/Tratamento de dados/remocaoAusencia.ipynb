```python
import pandas as pd

#1. Carregar os dados do arquivo

dadosExcel = pd.read_excel('dados_ausentes.xlsx')
#read excel pq o arquivo é um excel (pode ser json, cvs, etc). O valor entre '' é o caminho do arquivo

print(dadosExcel)
```

```python
#2. Verificar se há dados faltando
print(dadosExcel.isnull()) #isnull retorna em True os campos da tabela que estão vazio
```

```python
#3. Contar os valores ausentes
print(dadosExcel.isnull().sum())
```

```python
#4. Excluindo as linhas com dados ausentes

df_sem_nulos = dadosExcel.dropna() #Exclui as linhas que possuem valores vazios
print(df_sem_nulos)
```

```python
#5. Formata a tabela com valores ausentes
#Cidades ausentes - Desconhecido

dadosExcel['Cidade'].fillna('DESCONHECIDO', inplace = True) #Sempre que tiver 'na' no final da função, ele vai intervir no valores NaN
#'inplace = True' vai substituir os valores diretamente no data frame
#O arquivo de dados (excel, json, cvs, etc) original nunca será alterado, será mantido o original e será salvo uma nova versão do arquivo com os dados alterados

#Substituindo os valores vazios de salário por 0
dadosExcel['Salário'].fillna(0, inplace = True)

#Substituindo os valores vazios de idade pela média das idades
dadosExcel['Idade'].fillna(dadosExcel['Idade'].mean(), inplace = True) #O primeiro valor é a média resultada pela função .mean() na coluna Idade do data frame dadosExcel

print(dadosExcel)
```
