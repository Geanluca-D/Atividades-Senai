```python
import pandas as pd
import datetime
import time

df = pd.read_excel('despesas.xlsx')
df2 = pd.read_excel('orcamentos.xlsx')
```

```python
# Converte a data - despesas
df['Data'] = pd.to_datetime(df['Data'], unit='ms')
df['Data'] = df['Data'].dt.strftime('%d/%m/%Y')
display(df)
```

```python
# Apaga as linhas com valores vazios - despesas
df = df.dropna()
display(df)
```

```python
# Formatar a coluna Valor - despesas
df['Valor'] = df['Valor'].astype(float)
display(df)

df.to_excel('despesas-Tratado.xlsx', index=False)
```

```python
# Formatar os valores - orcamentos
df2['Valor_previsto'] =  df2['Valor_previsto'].astype(float).round(2)
df2['Valor_realizado'] = df2['Valor_realizado'].astype(float).round(2)

display(df2)
```

```python
# Mescla e formata as colunas mes e ano - orcamentos
df2['Mes'] = df2['Mes'].astype(str).str.zfill(2)
df2['Ano'] = df2['Ano'].astype(str)
df2['Data'] = pd.to_datetime(df2['Mes'] + '/' + df2['Ano'], format='%m/%Y')
df2['Data'] = df2['Data'].dt.strftime('%m/%y')
display(df2)

df2.to_excel('orcamentos-Tratado.xlsx', index=False)
```

```python
# Contabiliza células vazias
vazio = df2.isnull().sum()

print(vazio)
```
